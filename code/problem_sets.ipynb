<h3>If you want to keep exploring (evening / homework)</h3>

<p>
The analyses we worked through today are only a starting point. There are many ways to extend them, either by asking new scientific questions or by testing how robust your conclusions are to different analysis choices. Below are a few directions you can explore using the same dataset and methods.
</p>

<ol>

<li>
<p><strong>Decode context itself.</strong><br>
Instead of decoding the stimulus, ask whether neural activity contains information about the current task context. Can you predict whether the mouse is in the auditory or visual block? This can help reveal whether the brain maintains a representation of task context independently of the immediate sensory input.
</p>
</li>

<li>
<p><strong>Compare stimulus decoding across brain areas.</strong><br>
Run the same stimulus decoder separately for different recorded brain regions. Which areas provide the most information about stimulus identity? Does decoding accuracy change over time, and do different regions carry stimulus information at different stages of the trial?
</p>
</li>

<li>
<p><strong>Split trials by behavioral outcome.</strong><br>
Separate trials according to the mouse's behavior, for example hits versus misses, and repeat the decoding analysis. Is the stimulus represented more strongly on trials in which the mouse responds correctly? This can help distinguish neural representations that are simply stimulus-driven from those that may be related to successful task performance.
</p>
</li>

<li>
<p><strong>Examine catch trials.</strong><br>
Catch trials contain no stimulus, so they provide an interesting test of what the decoder has actually learned. What does the decoder predict when there is nothing to decode? Are its predictions biased toward a particular stimulus, context, or behavioral response?
</p>
</li>

<li>
<p><strong>Decode all four stimuli simultaneously.</strong><br>
Instead of performing pairwise decoding, train a multiclass decoder to distinguish <strong>vis1, vis2, sound1, and sound2</strong>. Inspect the resulting four-way confusion matrix. Which stimuli are most easily distinguished, and which are most frequently confused? Are errors more common within a sensory modality or across modalities?
</p>
</li>

<li>
<p><strong>Decode stimuli across different behavioral states.</strong><br>
If you have identified periods in which the mouse occupies different behavioral states, repeat the decoding analysis within each state. Does the amount of stimulus information in the neural population change depending on the mouse's behavioral state?
</p>
</li>

<li>
<p><strong>Explore additional features for characterizing behavioral state.</strong><br>
Behavioral state does not have to be defined using task performance alone. Consider incorporating measurements such as running speed, pupil size, facial movements, whisking, licking, or other available behavioral variables. Do these features reveal additional structure in the animal's behavior?
</p>
</li>

<li>
<p><strong>Ask whether behavioral states can be identified from neural activity.</strong><br>
So far, behavioral states may have been defined using behavioral measurements. Try reversing the question: if you apply state-space or clustering methods directly to neural activity, what kinds of states emerge? How closely do those neural states correspond to states defined from behavior?
</p>
</li>

<li>
<p><strong>Test the effect of spike-bin width.</strong><br>
The temporal bin size used to represent spikes is an analysis choice that can strongly influence the apparent structure of neural activity. Repeat an analysis using different bin widths, for example smaller or larger than the one used during the workshop. Which results remain stable, and which depend on temporal resolution?
</p>
</li>

<li>
<p><strong>Experiment with model choices and hyperparameters.</strong><br>
The methods introduced today contain many assumptions and tunable parameters. For the SVM, try different kernels or regularization strengths. For the HMM, experiment with the number of states or different observation distributions. Rather than looking only for the model with the highest performance, ask which conclusions remain consistent across reasonable modeling choices.
</p>
</li>

</ol>

<p>
The broader goal is to move beyond simply running an analysis pipeline and start asking how different choices change the scientific interpretation. Treat these directions as opportunities to generate new hypotheses, test robustness, and discover features of the dataset that we did not explicitly examine during the workshop.
</p>
