<h3>If you want to keep exploring (evening / homework)</h3>

<p>
These exercises build directly on the analyses from the workshop. They are opportunities to try variations of the methods, explore different parts of the dataset, and test how sensitive your results are to different analysis choices.
</p>

<ol>

<li>
<p><strong>Decode context itself.</strong><br>
Instead of decoding stimulus identity, ask whether neural activity contains information about the current task context. Can you predict whether the mouse is in the auditory or visual block?
</p>
</li>

<li>
<p><strong>Split trials by behavioral outcome.</strong><br>
Separate trials according to the mouse's behavior, for example hits versus misses, and repeat the decoding analysis. Does decoding performance differ depending on whether the mouse responded correctly?
</p>
</li>

<li>
<p><strong>Examine catch trials.</strong><br>
Catch trials contain no stimulus. What does a stimulus decoder predict on these trials? Is there any systematic structure in the decoder's predictions?
</p>
</li>

<li>
<p><strong>Decode all four stimuli simultaneously.</strong><br>
Instead of pairwise decoding, train a multiclass decoder to distinguish <strong>vis1, vis2, sound1, and sound2</strong>. Examine the four-way confusion matrix. Which stimuli are most easily distinguished, and which are most frequently confused? Are errors more common within or across sensory modalities?
</p>
</li>

<li>
<p><strong>Test the effect of spike-bin width.</strong><br>
Repeat one of the neural analyses using different spike-bin widths. Which results remain stable, and which depend on the temporal resolution used to represent neural activity?
</p>
</li>

<li>
<p><strong>Experiment with model choices and hyperparameters.</strong><br>
Explore some of the choices we made during the workshop. For the SVM, try different kernels or regularization strengths. For the HMM, try different numbers of states or observation distributions. How sensitive are your conclusions to these choices?
</p>
</li>

<li>
<p><strong>Explore additional features for characterizing behavioral state.</strong><br>
Try incorporating other available behavioral measurements, such as facial motion, pupil position, licking, or running. Do different combinations of features change the behavioral states you identify?
</p>
</li>

</ol>

<p>
The goal here is not necessarily to develop a complete new analysis, but to get comfortable modifying the approaches from the workshop and asking what happens when you change the question, the data, or the assumptions.
</p>

<h1>Possible Project Directions</h1>

<p>
The questions below are broader and could form the starting point for a more substantial project. They are intentionally open-ended: you can refine the question, combine multiple ideas, or develop a different analysis depending on what you find in the data.
</p>

<ol>

<li>
<p>
<strong>How does stimulus and context information vary across the brain?</strong><br>
Compare decoding across brain areas, recording sessions, and animals. Which areas consistently carry information about stimulus identity or task context? How much variability is there across sessions and animals, and what might explain that variability?
</p>
</li>

<li>
<p>
<strong>How is task-related information organized in population neural activity?</strong><br>
Explore low-dimensional representations of population activity using approaches such as PCA. How are stimulus, context, behavioral choice, and behavioral state represented in population activity? Do they organize neural activity along similar or different dimensions?
</p>
</li>

<li>
<p>
<strong>Does the temporal structure of spiking carry information beyond spike counts?</strong><br>
Most of our analyses represent neural activity using spike counts or firing rates. Explore whether features based on inter-spike intervals (ISIs) can also be used to decode stimulus identity. How does the information captured by spike timing compare with that captured by spike counts?
</p>
</li>

<li>
<p>
<strong>What determines how much a neuron contributes to population decoding?</strong><br>
Explore the properties of neurons that receive large decoder weights. Are decoder weights related to firing rate, stimulus selectivity, response variability, or other properties of individual neurons?
</p>
</li>

<li>
<p>
<strong>How is coordinated neural activity structured within and across brain areas?</strong><br>
Examine correlations between neurons within the same brain area and across different areas. Does correlation structure depend on stimulus, context, behavioral outcome, or behavioral state? How might these correlations relate to population decoding?
</p>
</li>

<li>
<p>
<strong>How does behavioral state influence neural representations?</strong><br>
Compare neural representations of the same stimulus across different behavioral states. Does the amount or structure of stimulus information change with the mouse's behavioral state? Are some brain areas more strongly affected by behavioral state than others?
</p>
</li>

<li>
<p>
<strong>Can behavioral state be identified from neural activity?</strong><br>
In the workshop, behavioral states were defined using behavioral measurements. Reverse the question and ask what states emerge when neural activity itself is used to identify states. How do these neural states relate to states identified from behavior?
</p>
</li>

</ol>

<p>
These directions are starting points rather than predefined projects. A good first step is to choose a question that interests you, make it more specific, and then decide which analyses and controls would be needed to answer it.
</p>
